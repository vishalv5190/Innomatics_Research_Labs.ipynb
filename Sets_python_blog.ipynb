# Sets in Python: How They Help Remove Duplicates and Improve Performance

This notebook explains Python Sets with examples and real-world use cases.

## Example 1: Creating a Set

```python
numbers = {1, 2, 3, 4}
print("Set:", numbers)
```

## Example 2: Removing Duplicates

```python
numbers = {1, 2, 2, 3, 4, 4}
print("Duplicates removed:", numbers)
```

## Example 3: Convert List to Set

```python
data = [10, 20, 20, 30, 40, 40]
unique_data = set(data)
print("Unique data:", unique_data)
```

## Example 4: Set Operations

```python
A = {1, 2, 3}
B = {3, 4, 5}
print("Union:", A.union(B))
print("Intersection:", A.intersection(B))
print("Difference:", A.difference(B))
```

## Example 5: Real‑World Example (Email Deduplication)

```python
emails = ['user@gmail.com', 'admin@gmail.com', 'user@gmail.com']
unique_emails = set(emails)
print("Unique emails:", unique_emails)
```

## Conclusion
Sets help remove duplicates and improve performance. They are useful in data cleaning and real-world applications.
